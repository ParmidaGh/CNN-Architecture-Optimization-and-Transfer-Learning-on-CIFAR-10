```python
from torch.nn.modules.dropout import Dropout
#مقدار دهی اولیه
batch_size = 128
learning_rate=0.001
model_name='خ'



import torch
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, random_split
import numpy as np
import os
import time


# محاسبه دقت
def calculate_accuracy(dataloader):
    correct = 0
    total = 0
    with torch.no_grad():
        for data in dataloader:
            images, labels = data
            images, labels = images.cuda(), labels.cuda()
            outputs = net(images)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    accuracy = round(100 * correct / total, 4)  # اعشار 4 رقمی
    return accuracy


#ساخت Loader
transform = transforms.Compose([transforms.ToTensor()])

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])  # نرمال سازی
])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True,
                                        download=True, transform=transform)

testset = torchvision.datasets.CIFAR10(root='./data', train=False,
                                       download=True, transform=transform)

#یک پنجم دیتا آموزش به صورت تصادفی به عنوان دیتا valid
trainset, validset = random_split(trainset,[45000,5000])
trainloader = torch.utils.data.DataLoader(trainset, batch_size=batch_size,
                                          shuffle=True, num_workers=2,pin_memory=True)
validloader = torch.utils.data.DataLoader(validset, batch_size=batch_size,
                                          shuffle=False, num_workers=2,pin_memory=True)
testloader = torch.utils.data.DataLoader(testset, batch_size=batch_size,
                                         shuffle=False, num_workers=2,pin_memory=True)

classes = ('plane', 'car', 'bird', 'cat',
           'deer', 'dog', 'frog', 'horse', 'ship', 'truck')


import torch.nn as nn
import torch.nn.functional as F


class Network(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, bias=True)
        self.batch_norm1 = nn.BatchNorm2d(32)
        self.pool1 = nn.MaxPool2d(2, stride=2)
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=32, kernel_size=3, bias=True)
        self.batch_norm2 = nn.BatchNorm2d(32)
        self.pool2 = nn.MaxPool2d(2, stride=2)
        self.conv3 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, bias=True)
        self.batch_norm3 = nn.BatchNorm2d(64)
        self.pool3 = nn.MaxPool2d(2, stride=2)
        self.conv4 = nn.Conv2d(in_channels=64, out_channels=64, kernel_size=3, padding=1, bias=True)
        self.batch_norm4 = nn.BatchNorm2d(64)
        self.pool4 = nn.MaxPool2d(2, stride=1)

        # Classification layers
        self.fc1 = nn.Linear(64 * 1 * 1, 256)
        self.dropout = nn.Dropout(p=0.3)  # Dropout layer
        self.fc2 = nn.Linear(256, 10)

        self.leaky_relu = nn.LeakyReLU(0.01)  # small negative slope

    def forward(self, x):
        x = self.leaky_relu(self.conv1(x))
        x = self.batch_norm1(x)
        x = self.pool1(x)
        x = self.leaky_relu(self.conv2(x))
        x = self.batch_norm2(x)
        x = self.pool2(x)
        x = self.leaky_relu(self.conv3(x))
        x = self.batch_norm3(x)
        x = self.pool3(x)
        x = self.leaky_relu(self.conv4(x))
        x = self.batch_norm4(x)
        x = self.pool4(x)
        x = torch.flatten(x, 1)  # flatten all dimensions except batch
        x = self.leaky_relu(self.fc1(x))
        x = self.dropout(x)  # Apply dropout
        x = self.fc2(x)
        return x

net = Network()
if torch.cuda.is_available():
  net = net.to(torch.device("cuda:0"))



import torch.optim as optim

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(net.parameters(), lr=learning_rate)


epoch=0
stop_condition=False

accuracy_list_valid=[]
error_list_train_plot=[]
error_list_valid_plot=[]


#Training part
start_time = time.time()
while (epoch < 20 and not stop_condition):
    for i, data in enumerate(trainloader, 0):

        # get the inputs; data is a list of [inputs, labels]
        inputs, labels = data
        inputs, labels = inputs.cuda(), labels.cuda()

        # zero the parameter gradients
        optimizer.zero_grad()

        # forward + backward + optimize
        outputs = net(inputs)
        #print(outputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()


    epoch+=1


    accuracy_valid=calculate_accuracy(validloader)
    accuracy_train=calculate_accuracy(trainloader)
    print(f'\nReport for epoch: {epoch}')
    print(f'valid accuracy: {accuracy_valid:.4f}, valid loss: {100-accuracy_valid:.4f}')
    print(f'train accuracy: {accuracy_train:.4f}, train loss: {100-accuracy_train:.4f}\n')


    error_list_valid_plot.append(100-accuracy_valid)
    error_list_train_plot.append(100-accuracy_train)
    accuracy_list_valid.append(accuracy_valid)

    #check stop condition
    last_3_acc = accuracy_list_valid[-3:]
    if max(last_3_acc) - sum(last_3_acc) / 3 <= 0.3:
        stop_condition = True
end_time = time.time()



#Report generating
avg_epoch_runtime=(end_time-start_time)/epoch

print("================================\n")
print(f"(Training finished in {epoch} epochs)")
print(f"Average epoch's runtime: {avg_epoch_runtime} seconds")
print(f'accuracy on test data: {calculate_accuracy(testloader):.4f}, loss of test data: {100-calculate_accuracy(testloader):.4f}\n\n')



import matplotlib.pyplot as plt

# Load a test image
dataiter = iter(testloader)
images, labels = next(dataiter)

single_image = images[2].unsqueeze(0)  # Selecting the first image and adding a batch dimension

# Display test image
plt.figure(figsize=(2, 2))
plt.imshow(np.transpose(images[2], (1, 2, 0)) * 0.5 + 0.5, extent=[0, 1, 0, 1])  # Convert from [-1, 1] to [0, 1]
plt.title(f"\nSelected Image - Label: {classes[labels[2].item()]}")
plt.axis('off')
plt.show()


# Pass image through network
channel = None
conv_layer = None

def get_channel_hook(layer):
    def hook(module, input, output):
        global channel
        channel = output.detach()
    return layer.register_forward_hook(hook)

def get_conv_layer(model):
    for module in model.children():
        if isinstance(module, nn.Conv2d):
            return module

conv_layer = get_conv_layer(net)
channel_hook = get_channel_hook(conv_layer)


net.eval()
with torch.no_grad():
    output = net(single_image.cuda())

channel_hook.remove()

# Get channel of first convolutional layer
channels = channel[0].cpu()

# Display feature maps (channels)
num_channels = channels.size(0)
plt.figure(figsize=(20, 8))
for i in range(num_channels):
    plt.subplot(4, 8, i+1)
    plt.imshow(channels[i], cmap='gray')
    plt.axis('off')
    plt.title(f'Channel {str(i+1)}')
plt.show()
```

```python
# Get weight tensor of first convolutional layer
weight = conv_layer.weight.data.cpu()

# Display filter weights (kernels)
num_filters = weight.size(0)
plt.figure(figsize=(20, 8))
for i in range(num_filters):
    plt.subplot(4, 8, i+1)
    plt.imshow(weight[i, 0], cmap='gray')
    plt.axis('off')
    plt.title(f'Kernel {str(i+1)}')
plt.show()
```

```python
# Set a fixed size for the images
image_height = 2
image_width = 4

num_show = min(num_channels, num_filters)
num_columns = 8
num_rows = -(-num_show // num_columns)  # Number of rows calculated dynamically

fig, axs = plt.subplots(num_rows, num_columns * 2, figsize=(24, image_height * num_rows))

for i in range(num_show):
    row_idx = i // num_columns
    col_idx = (i % num_columns) * 2

    # Display channel
    axs[row_idx, col_idx].imshow(channels[i], cmap='gray')
    axs[row_idx, col_idx].set_title(f'Channel {i+1}')
    axs[row_idx, col_idx].axis('off')

    # Display corresponding filter (kernel)
    axs[row_idx, col_idx + 1].imshow(weight[i, 0], cmap='gray')
    axs[row_idx, col_idx + 1].set_title(f'Kernel {i+1}')
    axs[row_idx, col_idx + 1].axis('off')

plt.tight_layout()
plt.show()
```
